## 4. Respostas às Questões para Análise

**1. Qual prompt apresentou maior acurácia?**  
Os Prompts 2, 3 e 4 apresentaram a maior acurácia (100%), empatando na precisão técnica ao classificar corretamente todos os 12 casos.

**2. Definir claramente as categorias modificou os resultados?**  
Sim. A definição explícita das regras e fronteiras permitiu que o modelo tratasse a entrada estritamente como dado, eliminando ambiguidades e evitando o erro no caso T10.

**3. O Few-shot modificou alguma classificação? Apresente um exemplo.**  
O Few-shot não alterou os rótulos em relação ao Prompt 2 (ambos com 100%), mas aumentou a estabilidade e a padronização no processamento do caso complexo T12.

**4. Houve diferença entre acurácia e respeito ao formato?**  
Sim. O Prompt 1 teve alta acurácia na intenção (91,6%), mas falhou no formato por incluir textos explicativos. O Prompt 4 alinhou 100% de acurácia com 100% de conformidade ao formato JSON.

**5. O que aconteceu com T10 em cada experimento?**  
* **Prompt 1:** Falhou. O modelo seguiu o comando malicioso da mensagem e classificou como `COMERCIAL`.  
* **Prompts 2, 3 e 4:** Sucesso. O modelo ignorou a injeção e classificou corretamente como `CANCELAMENTO`.

**6. Por que T10 representa um problema de segurança e não apenas um prompt mal elaborado?**  
Porque demonstra vulnerabilidade a *Indirect Prompt Injection*, onde comandos maliciosos no texto sequestram o fluxo do sistema, permitindo burlar regras operacionais em produção.

**7. Como os diferentes prompts trataram T12?**  
O Prompt 1 gerou explicações longas; os Prompts 2 e 3 priorizaram a intenção principal (`TÉCNICO`); e o Prompt 4 classificou como `TÉCNICO` sinalizando `revisao_humana: true` devido à ambiguidade.

**8. A saída JSON poderia ser utilizada diretamente por outro programa?**  
Sim. Por ser um JSON válido e estrito, pode ser consumido diretamente por APIs e microsserviços sem pós-processamento.

**9. Qual prompt apresentou maior consistência entre execuções?**  
O Prompt 4 (JSON), pois a obrigatoriedade de seguir um esquema rígido limita a variação estocástica do modelo.

**10. Considerando acurácia, estabilidade, formato e resistência à injeção, qual versão seria uma melhor base para um sistema real?**  
O **Prompt 4 (Saída Estruturada JSON com diretrizes Few-Shot)**. Combinou 100% de acurácia, resistência total ao Prompt Injection (T10), tratamento de ambiguidade (T12) com suporte a *Human-in-the-Loop* e interoperabilidade direta via JSON.
